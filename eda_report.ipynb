# EXPLORATORY DATA ANALYSIS (EDA)

This notebook performs data analysis on the credit card fraud dataset.

---

IMPORT LIBRARIES

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Class Distribution

This shows imbalance between fraud and non-fraud cases.

```python
sns.countplot(x='Class', data=data)
plt.title("Class Distribution")
plt.show()
```

## Boxplot Analysis

Used to detect outliers in transaction amount.

```python
sns.boxplot(x=data['Amount'])
plt.title("Amount Distribution")
plt.show()
```

## Correlation Heatmap

Shows relationship between features.

```python
plt.figure(figsize=(10,6))
sns.heatmap(data.corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

## PCA Visualization

Reduces dimensions to visualize data distribution.

```python
from sklearn.decomposition import PCA

X = data.drop('Class', axis=1)
y = data['Class']

pca = PCA(n_components=2)
pca_data = pca.fit_transform(X)

plt.scatter(pca_data[:,0], pca_data[:,1], c=y, cmap='coolwarm', alpha=0.5)
plt.title("PCA Visualization")
plt.show()
```

## EDA SUMMARY

- Dataset is highly imbalanced  
- Fraud cases are very rare  
- Data contains outliers  
- Features show weak correlation  
- PCA shows overlapping classes  

These issues justify the need for Generative AI.
